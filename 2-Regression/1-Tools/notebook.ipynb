```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn import datasets, linear_model, model_selection
```

```python
X, y = datasets.load_diabetes(return_X_y=True)
print(X.shape)
print(X[0])
```

```python
X = X[:, 2]
X = X.reshape((-1, 1))
```

## Split Training and testing data

```python
X_train, X_test, y_train, y_test = model_selection.train_test_split(
    X, y, test_size=0.33
)
```

## Linear Regression model training

```python
model = linear_model.LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

## Visualization

```python
plt.scatter(X_test, y_test, color="black")
plt.plot(X_test, y_pred, color="blue", linewidth=3)
plt.xlabel("Scaled BMIs")
plt.ylabel("Disease Progression")
plt.title("A Graph Plot Showing Diabetes Progression Against BMI")
plt.show()
```

```python
import pandas as pd
```

```python
linnerud = datasets.load_linnerud()

F = linnerud.data
t = linnerud.target

features_df = pd.DataFrame(data=F, columns=linnerud.feature_names)
targets_df = pd.DataFrame(data=t, columns=linnerud.target_names)

print("Features:")
print(features_df.head())
print("")
print("Targets:")
print(targets_df.head())
```

```python
F = F[:, 0]
F = F.reshape((-1, 1))

t = t[:, 0]
t = t.reshape((-1, 1))
```

```python
X_train, X_test, y_train, y_test = model_selection.train_test_split(
    F, t, test_size=0.33
)
```

```python
model = linear_model.LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

```python
plt.scatter(X_test, y_test, color="black")
plt.plot(X_test, y_pred, color="blue", linewidth=3)
plt.xlabel("Excercise Amount")
plt.ylabel("Weight")
plt.title("A Graph Plot Showing Excercise Data Compared To Weight")
plt.show()
```
